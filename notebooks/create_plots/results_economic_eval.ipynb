```python
class CustomUnpickler(pickle.Unpickler):
    def find_class(self, module, name):
        try:
            return super().find_class(module, name)
        except (AttributeError, ModuleNotFoundError):
            # Return a dummy class for missing classes
            return type(name, (), {})


for file_name in file_names:
    with open(
        f"../evaluate_models/results/economical_evaluation/{file_name}_trading.pkl",
        "rb",
    ) as f:
        models_dict.extend(CustomUnpickler(f).load())
    with open(
        f"../evaluate_models/results/metric_evaluation/{file_name}.pkl",
        "rb",
    ) as f:
        models_dict_point_interval.extend(CustomUnpickler(f).load())
```

```python
# Load the data
val_date = "2022-12-01"
test_date = "2023-12-01"
end_date = "2024-11-30"
data_file_path = "../../data/processed/smard_data_201810010000_202501010000.npz"
data_module = EPFDataModule(
    data_file_path=data_file_path,
    val_date=val_date,
    test_date=test_date,
    end_date=end_date,
    batch_size=32,
    standardization_case=standardization_case,
)

train_input, train_labels = data_module.train_dataset[:]
val_input, val_labels = data_module.val_dataset[:]
test_input, test_labels = data_module.test_dataset[:]

data_input, data_labels = test_input, test_labels

data_labels = data_labels * data_module.scale_target + data_module.offset_target
data_labels = data_labels.detach().numpy()
```

```python
optimal_bids = trading_optimal_bids(data_labels)
optimal_bids_profit = np.sum(optimal_bids["profit_order"]).flatten()
optimal_bids_profit_per_day = np.sum(
    np.array(optimal_bids["profit_order"]).reshape(-1, 2), axis=1
)
optimal_bids_ptp = np.mean(optimal_bids["profit_order"]).flatten()
print(optimal_bids_profit)
print(optimal_bids_ptp)
```

```python
fixed_time = trading_fixed_hours(
    data_labels,
    (val_labels * data_module.scale_target + data_module.offset_target)
    .detach()
    .numpy(),
)
fixed_time_profit = np.sum(fixed_time["profit_order"]).flatten()
fixed_time_profit_per_day = (
    np.array(fixed_time["profit_order"])[: int(len(fixed_time["profit_order"]) / 2)]
    + np.array(fixed_time["profit_order"])[int(len(fixed_time["profit_order"]) / 2) :]
)
fixed_time_ptp = np.mean(fixed_time["profit_order"]).flatten()
print(fixed_time_profit)
print(fixed_time_ptp)
```
